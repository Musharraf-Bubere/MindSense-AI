# Missing Value Handling

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

missing_values.sort_values(ascending=False)
```

```python
pd.crosstab(
    df["Working Professional or Student"],
    df["Academic Pressure"].isnull()
)
```

```python
df["Academic Pressure"].value_counts(dropna=False).sort_index()
```

The missing values in the following columns are logical (structural) missing values rather than data collection errors.

Student-related features:
- Academic Pressure
- CGPA
- Study Satisfaction

These are missing for working professionals because they are not applicable.

Working professional-related features:
- Work Pressure
- Job Satisfaction

These are missing for students because they are not applicable.

Since multiple machine learning models will be evaluated and several of them do not support NaN values, these logical missing values will be replaced with -1, which acts as a sentinel value representing "Not Applicable".

```python
logical_missing_cols = [
    "Academic Pressure",
    "CGPA",
    "Study Satisfaction",
    "Work Pressure",
    "Job Satisfaction"
]

df[logical_missing_cols] = df[logical_missing_cols].fillna(-1)
```
