```python
import pandas as pd
import joblib

# Load model
model = joblib.load("../Models/logistic_regression.pkl")
feature_columns = joblib.load("../Models/feature_columns.pkl")

# Create one sample with all feature columns
sample = pd.DataFrame([[0] * len(feature_columns)], columns=feature_columns)

prediction = model.predict(sample)[0]
probability = model.predict_proba(sample)[0]

print(f"Prediction : {'Yes' if prediction == 1 else 'No'}")
print(f"Confidence : {probability[prediction] * 100:.2f}%")
```
